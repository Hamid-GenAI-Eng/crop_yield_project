```python
fao_pak = fao_df[fao_df['Area'] == 'Pakistan'].copy()
```

```python
print(fao_pak.info())
```

```python
years_cols = [c for c in fao_pak.columns if c.startswith('Y')]
```

```python
id_vars = ['Area', 'Item', 'Element']
```

```python
fao_long = fao_pak.melt(id_vars=id_vars, value_vars=years_cols, var_name='Year_str', value_name='Value')
```

```python
fao_long['Year'] = fao_long['Year_str'].str.replace('Y', '').astype(int)
```

```python
fao_pivot = fao_long.pivot_table(
    index=['Year', 'Item'],
    columns='Element',
    values='Value'
).reset_index()
```

```python
fao_pivot.columns.name = None
```

```python
major_crops = ['Wheat', 'Rice, paddy', 'Maize', 'Sugarcane', 'Cotton seed', 'Cotton lint']
fao_final = fao_pivot[fao_pivot['Item'].isin(major_crops)].copy()
```

```python
weather_long = weather_df.melt(
    id_vars=['PARAMETER', 'YEAR'],
    value_vars=['JAN', 'FEB', 'MAR', 'APR', 'MAY', 'JUN', 'JUL', 'AUG', 'SEP', 'OCT', 'NOV', 'DEC'],
    var_name='Month',
    value_name='Value'
)
```

```python
month_map = {
    'JAN': 'Q1', 'FEB': 'Q1', 'MAR': 'Q1',
    'APR': 'Q2', 'MAY': 'Q2', 'JUN': 'Q2',
    'JUL': 'Q3', 'AUG': 'Q3', 'SEP': 'Q3',
    'OCT': 'Q4', 'NOV': 'Q4', 'DEC': 'Q4'
}
weather_long['Quarter'] = weather_long['Month'].map(month_map)
```

```python
agg_rules = {
    'PRECTOTCORR': 'sum',
    'T2M_MAX': 'mean',
    'T2M_MIN': 'mean',
    'RH2M': 'mean',
    'ALLSKY_SFC_SW_DWN': 'mean'
}
```

```python
weather_feats = []

for param, agg_func in agg_rules.items():
    # Filter for specific parameter
    subset = weather_long[weather_long['PARAMETER'] == param]
    
    # Group by Year and Quarter
    grouped = subset.groupby(['YEAR', 'Quarter'])['Value'].agg(agg_func).reset_index()
    
    # Pivot to get Year | Q1_Param | Q2_Param ...
    pivoted = grouped.pivot(index='YEAR', columns='Quarter', values='Value')
    pivoted.columns = [f"{param}_{col}" for col in pivoted.columns]
    
    weather_feats.append(pivoted)
```

```python
annual_feats = weather_df.pivot(index='YEAR', columns='PARAMETER', values='ANN')
annual_feats.columns = [f"{col}_ANN" for col in annual_feats.columns]
weather_feats.append(annual_feats)
```

```python
weather_final = pd.concat(weather_feats, axis=1).reset_index()
weather_final.rename(columns={'YEAR': 'Year'}, inplace=True)
```

```python
merged_df = pd.merge(fao_final, weather_final, on='Year', how='inner')
```

```python
print("Merged Data Head:")
print(merged_df.head())
```

```python
print("\nMerged Data Info:")
print(merged_df.info())
```

```python
merged_df.to_csv('../data/processed/merged_crop_weather_data.csv', index=False)
print("\nSaved 'merged_crop_weather_data.csv'")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
wheat_df = merged_df[merged_df['Item'] == 'Wheat'].copy()
```

```python
wheat_df = wheat_df.sort_values('Year')
wheat_df['Yield_Lag1'] = wheat_df['Yield'].shift(1)
```

```python
wheat_df['Q1_Heat_Index'] = wheat_df['T2M_MAX_Q1'] * wheat_df['RH2M_Q1'] # Humid Heat
wheat_df['Q1_Rain_Temp_Ratio'] = wheat_df['PRECTOTCORR_Q1'] / (wheat_df['T2M_MAX_Q1'] + 0.1)
```

```python
cols_to_corr = [
    'Yield', 
    'Yield_Lag1',
    'PRECTOTCORR_Q1', 'T2M_MAX_Q1', 'T2M_MIN_Q1', # Winter/Spring Weather
    'PRECTOTCORR_Q4', 'T2M_MAX_Q4',               # Sowing Season Weather (Oct-Dec)
    'Q1_Heat_Index'
]
```

```python
corr_matrix = wheat_df[cols_to_corr].corr()
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", center=0)
plt.title('Correlation Matrix: Wheat Yield vs Seasonal Weather (Pakistan)')
plt.savefig('../data/processed/wheat_correlation_heatmap.png')
```

```python
plt.figure(figsize=(10, 5))
plt.plot(wheat_df['Year'], wheat_df['Yield'], marker='o', linestyle='-')
plt.title('Wheat Yield in Pakistan (2000-2023)')
plt.xlabel('Year')
plt.ylabel('Yield (hg/ha)')
plt.grid(True)
plt.savefig('../data/processed/wheat_yield_trend.png')

print("EDA Analysis Complete.")
print(wheat_df[['Year', 'Yield', 'Yield_Lag1', 'Q1_Heat_Index']].head())
```
